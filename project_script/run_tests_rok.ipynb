```python
import pandas as pd
import numpy as np

# Make sure you have the column names (expanded_col_names or header_expanded_train)
feature_names = expanded_col_names  # same order used in training

# Get coefficients
coefs = model.coef_.flatten()

# Build a DataFrame for easier analysis
coef_df = pd.DataFrame({
    "feature": feature_names,
    "coefficient": coefs,
    "abs_coeff": np.abs(coefs)
}).sort_values("abs_coeff", ascending=False)

print(coef_df.head(20))  # Top 20 most influential features

```

```python
import matplotlib.pyplot as plt

top_n = 25

top_features = coef_df.head(top_n).sort_values("coefficient")

plt.figure(figsize=(8, 6))
plt.barh(top_features["feature"], top_features["coefficient"], color=np.sign(top_features["coefficient"]).map({1: "green", -1: "red"}))
plt.xlabel("Coefficient (log-odds weight)")
plt.title(f"Top {top_n} Most Influential Features")
plt.axvline(0, color='k', linestyle='--')
plt.tight_layout()
plt.show()
```
